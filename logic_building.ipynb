```python
# largest number in list

numbers = [3,6,8,9,11]
largest = numbers[0]

for number in numbers:
  if number > largest:
    largest = number
print(largest)
```

```python
# if statements

numbers = [4, 9, 2, 7, 5]
target = 12

found = False

for number in numbers:
    if number == target:
        found = True
        break

if found:
    print("Found it!")
else:
    print("Not found")
```

```python
# find smallest and largest both in one loop

numbers = [4, 9, 2, 7, 5]

smallest = numbers[0]
largest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

    if number > largest:
        largest = number

print("Smallest:", smallest)
print("Largest:", largest)
```

```python
# position finding algorithm
def find_index(numbers, target):
    for i in range(len(numbers)):
        if numbers[i] == target:
            return i

    return -1

print(find_index([10, 20, 30, 40], 30))
print(find_index([10, 20, 30, 40], 99))

# Python also gives us a convenient way to get both the index and the value

numbers = [10, 20, 30, 40]

for i, number in enumerate(numbers):
    print(i, number)
```

```python
# O(n²) — nested loops

numbers = [1, 2, 3, 4]

for a in numbers:
    for b in numbers:
        print(a, b)
```

```python
# Write a function called count_vowels that counts how many vowels are in a word. The vowels are a, e, i, o, and u.

def count_vowels(word):
  vowels = "aeiouAEIOU"
  count = 0
  for char in word:
    if char in vowels:
      count += 1
  return count

print(count_vowels("banana"))
```

```python
# Reversing a string function

def reverse_string(word):
    reversed_word = ""

    for char in word:
        reversed_word = char + reversed_word

    return reversed_word


print(reverse_string("hello"))
print(reverse_string("python"))

# Python shortcut for reversing string
word = "hello"
print(word[::-1])

# Palindrome function

def is_palindrome(word):
    reversed_word = reverse_string(word)

    if word == reversed_word:
        return True
    else:
        return False


print(is_palindrome("madam"))
print(is_palindrome("hello"))

# Python shortcut for checking palindrome words

def is_palindromeshortcut(word):
    return word == word[::-1]

print(is_palindromeshortcut("racecar"))
print(is_palindromeshortcut("python"))
print(is_palindromeshortcut("level"))
```

```python
# Two pointers - to check palindrome

def is_palindrome(word):
    left = 0
    right = len(word) - 1

    while left < right:
        if word[left] != word[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome("racecar"))
```
